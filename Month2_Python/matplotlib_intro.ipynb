```python
import matplotlib.pyplot as plt
```

```python
students = ["Prathamesh", "Khushal", "Medha", "Mohit", "Kimaya", "Rucha"]
marks = [90, 85, 77, 80, 88, 76]

plt.plot(students, marks)
plt.title("Students marks analysis")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
plt.bar(students, marks)
plt.title("Students marks analysis")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
plt.barh(students, marks)
plt.title("Students marks analysis")
plt.show()
```

```python
cities = ["Pune","Mumbai","Delhi"]
counts = [3,2,1]

plt.pie(counts, labels=cities, autopct='%1.1f%%')
plt.title("Student Distribution by City")
plt.show()
```

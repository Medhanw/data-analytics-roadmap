```python
import matplotlib.pyplot as plt
```

```python
students = ["Prathamesh", "Khushal", "Medha", "Mohit", "Kimaya", "Rucha"]
marks = [90, 85, 77, 80, 88, 76]

plt.plot(students, marks)
plt.title("Students marks analysis")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
plt.bar(students, marks)
plt.title("Students marks analysis")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
plt.barh(students, marks)
plt.title("Students marks analysis")
plt.show()
```

```python
cities = ["Pune","Mumbai","Delhi"]
counts = [3,2,1]

plt.pie(counts, labels=cities, autopct='%1.1f%%')
plt.title("Student Distribution by City")
plt.show()
```

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Name" : ["Prathamesh", "Khushal", "Medha", "Mohit", "Kimaya", "Rucha"],
    "Marks" : [90, 85, 77, 80, 88, 76],
    "City" : ["Pune", "Mumbai", "Banglore", "Hydrabad", "Ahmdabad", "Chennai"]

}

df = pd.DataFrame(data)

df
```

```python
sns = sns.barplot(x="Name", y="Marks", data=df)
plt.title("Students marks analysis")
plt.show()
```

```python
sns.countplot(x="City", data=df)

plt.title("Students per City")

plt.show()
```

```python
sns.boxplot(y="Marks", data=df)
plt.title("Marks Distribution")
plt.show()
```

```python
sns.histplot(df["Marks"], bins=5)

plt.title("Marks Distribution")

plt.show()
```

```python
students = [
    {"name": "Alisha", "marks": 78},
    {"name": "Ayesha", "marks": 94},
    {"name": "Nandini", "marks": 69},
    {"name": "Selina", "marks": 55},
    {"name": "Disha", "marks": 80},
    {"name": "Smruti", "marks": 90},
    {"name": "Suha", "marks": 61},
    {"name": "Tanisha", "marks": 67},
    {"name": "Janhavi", "marks": 45},
    {"name": "Sayali", "marks": 97},
]
```

```python
for s in students:
    print(s["name"], "-", s["marks"])
```

```python
total = 0
for s in students:
    total += s["marks"]

average = total / len(students)
print("Average marks: ", average)
```

```python
highest = 0
top_student = ""
for s in students:
    if s["marks"] > highest:
        highest = s["marks"]
        top_student = s["name"]

print("Top Students: ", top_student, "-", highest)
```

```python
pass_count = 0
fail_count = 0

for s in students:
    if s["marks"] >= 60:
        pass_count += 1
    else:
        fail_count += 1

print("Pass: ", pass_count)
print("Fail: ", fail_count)
```

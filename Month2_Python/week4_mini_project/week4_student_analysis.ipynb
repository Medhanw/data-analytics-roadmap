```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = {
    "Name" : ["Prathamesh", "Khushal", "Medha", "Mohit", "Kimaya", "Rucha"],
    "Marks" : [85, 90, 78, 92, 88, 80],
    "City" :["Mumbai", "Delhi", "Bangalore", "Chennai", "Kolkata", "Hyderabad"]
}
df = pd.DataFrame(data)

df
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df["Marks"].mean() 
```

```python
df["Marks"].max()
```

```python
df["City"].value_counts()
```

```python
df.groupby("City")["Marks"].mean()
```

```python
df.groupby("City")["Marks"].agg(["mean", "max", "min", "count"])
```

```python
sns.barplot(x="City", y="Marks", data=df)
plt.title("Average students marks by city")
plt.show()
```

```python
sns.histplot(df["Marks"], bins=5)
plt.title("Marks distribution")
plt.show()
```

```python
sns.boxplot(y="Marks", data=df)
plt.title("Marks Spread")
plt.show()
```
